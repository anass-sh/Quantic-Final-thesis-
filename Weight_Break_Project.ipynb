```python
import pandas as pd
import numpy as np
import re
from pathlib import Path

pd.set_option("display.max_columns", 200)

def clean_cols(df):
    df = df.copy()
    df.columns = [re.sub(r"\s+", "_", str(c).strip().upper()) for c in df.columns]
    return df

def assert_path(p, key):
    if p is None:
        raise ValueError(f"FILES['{key}'] is None -> file picker did not find a match. Check filenames in DATA_DIR.")
    p = Path(p)
    if not p.exists():
        raise FileNotFoundError(f"FILES['{key}'] points to a path that does not exist:\n{p}")
    return str(p)

def load_oracle_tm(path, site):
    df = pd.read_excel(path, engine="openpyxl")  # explicit engine for .xlsx
    df = clean_cols(df)
    df["SITE"] = site
    for c in ["BEGINNING_WEIGHT", "ENDING_WEIGHT"]:
        if c in df.columns:
            df[c] = pd.to_numeric(df[c], errors="coerce")
    return df

# --- checks ---
if "FILES" not in globals():
    raise NameError("FILES is not defined in this notebook. Run your FILES picker cell first.")

oracle_pdc_path = assert_path(FILES.get("oracle_pdc"), "oracle_pdc")
oracle_qmp_path = assert_path(FILES.get("oracle_qmp"), "oracle_qmp")

print("oracle_pdc ->", oracle_pdc_path)
print("oracle_qmp ->", oracle_qmp_path)

# --- load ---
pdc = load_oracle_tm(oracle_pdc_path, "PDC")
qmp = load_oracle_tm(oracle_qmp_path, "QMP")

oracle = pd.concat([pdc, qmp], ignore_index=True)
print("Oracle rows:", oracle.shape[0], "cols:", oracle.shape[1])
oracle.head(3)
```

Find “candidate” sheets that likely contain rate tables

This heuristic looks for keywords like ZONE / RATE / KG / WEIGHT in the first ~30 rows.

```python
KEYWORDS = ["ZONE", "RATE", "RATES", "KG", "KGS", "WEIGHT", "EXPORT", "IMPORT", "ECONOMY", "EXPRESS"]

def sheet_keyword_score(path, sheet, nrows=30):
    df = pd.read_excel(path, sheet_name=sheet, header=None, nrows=nrows)
    text = " ".join(df.astype(str).fillna("").values.flatten()).upper()
    score = sum(1 for k in KEYWORDS if k in text)
    return score

dhl_scores = []
for sh in pd.ExcelFile(dhl_path).sheet_names:
    dhl_scores.append((sh, sheet_keyword_score(dhl_path, sh)))

dhl_scores = sorted(dhl_scores, key=lambda x: x[1], reverse=True)
print("Top candidate sheets:")
for sh, sc in dhl_scores[:10]:
    print(f"{sc:02d}  {sh}")
```

Auto-detect the header row for each candidate sheet

Many carrier files have a few “title rows” before the real table header.

```python
def detect_header_row(path, sheet, max_scan_rows=40):
    df = pd.read_excel(path, sheet_name=sheet, header=None, nrows=max_scan_rows)
    # look for a row that contains likely column headers
    # (zone, weight, kg, etc.)
    for i in range(min(max_scan_rows, len(df))):
        row = df.iloc[i].astype(str).str.upper().fillna("")
        row_text = " ".join(row.values)
        if ("ZONE" in row_text or "WEIGHT" in row_text or "KG" in row_text) and (row.nunique() > 3):
            return i
    return None

candidates = [sh for sh, sc in dhl_scores if sc >= 3][:12]  # adjust threshold if needed
print("Candidate sheets considered:", candidates)

header_map = {}
for sh in candidates:
    header_row = detect_header_row(dhl_path, sh)
    header_map[sh] = header_row
    print(f"{sh:40s} header_row={header_row}")
```

Read candidate sheets using detected headers (clean columns)

```python
def read_with_header(path, sheet, header_row):
    df = pd.read_excel(path, sheet_name=sheet, header=header_row)
    df.columns = clean_cols(df.columns)
    df["__SOURCE_FILE"] = Path(path).name
    df["__SOURCE_SHEET"] = sheet
    df["__HEADER_ROW"] = header_row
    return df

dhl_tables = []
for sh, header_row in header_map.items():
    if header_row is None:
        continue
    try:
        t = read_with_header(dhl_path, sh, header_row)
        # drop fully empty columns
        t = t.dropna(axis=1, how="all")
        dhl_tables.append(t)
        print("Loaded:", sh, "| shape:", t.shape)
    except Exception as e:
        print("Skip:", sh, "| error:", e)

print("Loaded tables:", len(dhl_tables))
```

Standardize wide “zone x weight columns” tables (common DHL format)



```python
import re
import numpy as np
import pandas as pd

def normalize_header_token(x):
    s = str(x).strip().upper()
    s = s.replace(",", ".")                # 0,5 -> 0.5
    s = re.sub(r"\s+", " ", s)
    s = s.replace("KG", "").replace("KGS", "").strip()
    return s

def parse_weight_header(x):
    """
    Returns (weight_value_float) if header looks like a single weight,
    else returns None.
    Handles: '0,5', '0.5', '0.5 KG', ' 10 '.
    """
    s = normalize_header_token(x)
    # keep only numeric-ish patterns
    if re.fullmatch(r"\d+(\.\d+)?", s):
        return float(s)
    return None

def parse_weight_range_cell(x):
    """
    Returns (low, high) if a cell looks like a weight range like '0-0.5' or '0.0 - 0.5',
    else None.
    """
    s = str(x).strip().upper().replace(",", ".")
    s = re.sub(r"\s+", "", s)
    # examples: "0-0.5", "0.0-0.5", "10-20"
    m = re.fullmatch(r"(\d+(\.\d+)?)-(\d+(\.\d+)?)", s)
    if m:
        return float(m.group(1)), float(m.group(3))
    return None

def infer_orientation(df):
    """
    Decide if weights are columns or rows.
    - weights-as-columns: many column names parse as weights
    - weights-as-rows: first column contains numeric weights / ranges and other columns look like zones
    """
    cols = list(df.columns)
    w_cols = [c for c in cols if parse_weight_header(c) is not None]
    if len(w_cols) >= 3:
        return "weights_as_columns"

    # check weights as first column values
    first = df.columns[0]
    s = df[first].astype(str).head(30).tolist()
    n_numeric = 0
    n_range = 0
    for v in s:
        v2 = str(v).strip().replace(",", ".")
        if re.fullmatch(r"\d+(\.\d+)?", v2):
            n_numeric += 1
        if parse_weight_range_cell(v) is not None:
            n_range += 1
    if n_numeric + n_range >= 5:
        return "weights_as_rows"
    return "unknown"

def wide_weights_cols_to_long(df):
    # zone col = first non-provenance col that is NOT a weight header
    prov = {"__SOURCE_FILE","__SOURCE_SHEET","__HEADER_ROW"}
    nonprov_cols = [c for c in df.columns if c not in prov]
    weight_cols = [c for c in nonprov_cols if parse_weight_header(c) is not None]
    if len(weight_cols) < 3:
        return None

    zone_col = [c for c in nonprov_cols if c not in weight_cols][0]  # usually "ZONE"
    out = df[[zone_col] + weight_cols + [c for c in df.columns if c in prov]].copy()
    out = out.rename(columns={zone_col: "ZONE"})
    out = out.melt(
        id_vars=["ZONE"] + [c for c in df.columns if c in prov],
        value_vars=weight_cols,
        var_name="WEIGHT",
        value_name="RATE"
    )
    out["WEIGHT"] = out["WEIGHT"].map(parse_weight_header)
    out["RATE"] = pd.to_numeric(out["RATE"], errors="coerce")
    out = out.dropna(subset=["ZONE","WEIGHT","RATE"])
    return out

def wide_weights_rows_to_long(df):
    """
    Common DHL layout: first column = weight (or weight range), columns = zones.
    We'll melt zones into long format.
    """
    prov = {"__SOURCE_FILE","__SOURCE_SHEET","__HEADER_ROW"}
    nonprov_cols = [c for c in df.columns if c not in prov]

    weight_col = nonprov_cols[0]  # first column has weights
    zone_cols = nonprov_cols[1:]

    out = df[[weight_col] + zone_cols + [c for c in df.columns if c in prov]].copy()
    out = out.rename(columns={weight_col: "WEIGHT_RAW"})

    out = out.melt(
        id_vars=["WEIGHT_RAW"] + [c for c in df.columns if c in prov],
        value_vars=zone_cols,
        var_name="ZONE",
        value_name="RATE"
    )

    # Parse WEIGHT_RAW into WEIGHT (single) or WEIGHT_FROM/TO
    out["RATE"] = pd.to_numeric(out["RATE"], errors="coerce")

    # try single weight first
    w_single = out["WEIGHT_RAW"].astype(str).str.strip().str.replace(",", ".", regex=False)
    out["WEIGHT"] = pd.to_numeric(w_single, errors="coerce")

    # ranges
    ranges = out["WEIGHT_RAW"].map(parse_weight_range_cell)
    out["WEIGHT_FROM"] = ranges.map(lambda x: x[0] if isinstance(x, tuple) else np.nan)
    out["WEIGHT_TO"]   = ranges.map(lambda x: x[1] if isinstance(x, tuple) else np.nan)

    out = out.dropna(subset=["ZONE","RATE"], how="any")

    return out

standardized = []
debug = []

for t in dhl_tables:
    orient = infer_orientation(t)
    sheet = t["__SOURCE_SHEET"].iloc[0]
    debug.append((sheet, orient, t.shape, list(t.columns)[:10]))

    if orient == "weights_as_columns":
        out = wide_weights_cols_to_long(t)
        if out is not None and len(out):
            out["PARSER"] = "weights_as_columns"
            standardized.append(out)

    elif orient == "weights_as_rows":
        out = wide_weights_rows_to_long(t)
        if out is not None and len(out):
            out["PARSER"] = "weights_as_rows"
            standardized.append(out)

print("Classification summary:")
pd.DataFrame(debug, columns=["sheet","orientation","shape","first_cols"]).head(15)
```

build proper ranges

```python
import numpy as np

df = dhl_rate_fact.copy()
df = df[df["ZONE"].astype(str).str.contains("UNNAMED", na=False) == False].copy()

# Ensure numeric breakpoints
df["WEIGHT"] = pd.to_numeric(df["WEIGHT"], errors="coerce")
df["RATE"] = pd.to_numeric(df["RATE"], errors="coerce")
df = df.dropna(subset=["ZONE","WEIGHT","RATE"])

group_cols = [c for c in ["SERVICE_LEVEL","ZONE"] if c in df.columns]

def bands_half_open(g):
    g = g.sort_values("WEIGHT").copy()

    # Unique weights only (duplicate columns can create repeats)
    g = g.drop_duplicates(subset=["WEIGHT"])

    # half-open: [prev, current)
    g["WEIGHT_FROM"] = g["WEIGHT"].shift(1).fillna(0)
    g["WEIGHT_TO"] = g["WEIGHT"]

    # drop invalid
    g = g[g["WEIGHT_TO"] > g["WEIGHT_FROM"]]
    return g

df3 = df.groupby(group_cols, dropna=False, group_keys=False).apply(bands_half_open)

print("df3 shape:", df3.shape)
display(df3.head(20))
```

Sanity check, verifying that there are no negative ranges and they’re ordered.

```python
import numpy as np
import pandas as pd

# df3 is your final banded table
# Must include WEIGHT_FROM and WEIGHT_TO

group_cols = [c for c in ["SERVICE_LEVEL", "ZONE"] if c in df3.columns]
print("Continuity check grouping by:", group_cols)

def continuity_gaps(g):
    g = g.sort_values(["WEIGHT_FROM","WEIGHT_TO"]).copy()
    return (g["WEIGHT_FROM"].iloc[1:].values - g["WEIGHT_TO"].iloc[:-1].values)

chk = (df3.groupby(group_cols, dropna=False)
          .apply(continuity_gaps)
          .reset_index(name="GAPS"))

display(chk.head(15))

# Flags
def has_pos(arr):
    return np.any(np.array(arr) > 1e-9) if hasattr(arr, "__len__") else False

def has_neg(arr):
    return np.any(np.array(arr) < -1e-9) if hasattr(arr, "__len__") else False

chk["HAS_POS_GAP"] = chk["GAPS"].apply(has_pos)
chk["HAS_NEG_GAP"] = chk["GAPS"].apply(has_neg)

print("Groups with positive gaps:", chk["HAS_POS_GAP"].sum())
print("Groups with overlaps (negative gaps):", chk["HAS_NEG_GAP"].sum())

# show only problematic groups
display(chk[chk["HAS_POS_GAP"] | chk["HAS_NEG_GAP"]].head(30))
```

carrier | direction | origin | destination | service_level | lane | weight_from | weight_to | weight_uom | rate | currency | source_file | source_sheet

Because Nunner files can be structured in different ways (country tabs vs lane tabs), we’ll do this in a robust “analytical” way:

inventory sheets

detect which sheets contain actual rate tables

auto-detect header rows

parse either “weights as rows” or “weights as columns”

```python
import pandas as pd
import re
import numpy as np

nunner_out_path = FILES["nunner_out"]
print("Nunner OUT file:", nunner_out_path)

xls = pd.ExcelFile(nunner_out_path)
print("Sheets:", len(xls.sheet_names))
print("First 25 sheet names:")
for sh in xls.sheet_names[:25]:
    print("-", sh)

KEYWORDS = ["RATE", "RATES", "EUR", "EURO", "KG", "KGS", "WEIGHT", "DEST", "COUNTRY", "ZONE", "LANE"]

def sheet_score(path, sheet, nrows=40):
    df = pd.read_excel(path, sheet_name=sheet, header=None, nrows=nrows)
    text = " ".join(df.astype(str).fillna("").values.flatten()).upper()
    return sum(1 for k in KEYWORDS if k in text)

scores = [(sh, sheet_score(nunner_out_path, sh)) for sh in xls.sheet_names]
scores = sorted(scores, key=lambda x: x[1], reverse=True)

print("\nTop candidate sheets:")
for sh, sc in scores[:15]:
    print(f"{sc:02d}  {sh}")
```

Strong header-row detection + load candidate tables

```python
def best_header_row(path, sheet, scan_rows=80):
    df = pd.read_excel(path, sheet_name=sheet, header=None, nrows=scan_rows)
    best = (None, -1)
    for i in range(min(scan_rows, len(df))):
        row = df.iloc[i].astype(str).str.upper().fillna("")
        row_text = " ".join(row.values)

        # score by presence of business headers
        score = 0
        for k in ["WEIGHT", "KG", "DEST", "DESTINATION", "COUNTRY", "ZONE", "LANE", "RATE", "EUR"]:
            if k in row_text:
                score += 3

        # prefer rows with many non-empty cells
        nonempty = (row != "NAN").sum()
        score += min(nonempty, 12) * 0.2

        if score > best[1]:
            best = (i, score)
    return best

def clean_cols(cols):
    return [re.sub(r"\s+", "_", str(c).strip().upper()) for c in cols]

candidate_sheets = [sh for sh, sc in scores if sc >= 3][:20]  # threshold can be adjusted
print("Candidate sheets:", len(candidate_sheets))

nunner_tables = []
meta = []

for sh in candidate_sheets:
    hr, sc = best_header_row(nunner_out_path, sh)
    meta.append((sh, hr, sc))
    if hr is None:
        continue
    try:
        df = pd.read_excel(nunner_out_path, sheet_name=sh, header=hr)
        df.columns = clean_cols(df.columns)
        df = df.dropna(axis=1, how="all")
        df["__SOURCE_FILE"] = Path(nunner_out_path).name
        df["__SOURCE_SHEET"] = sh
        df["__HEADER_ROW"] = hr
        nunner_tables.append(df)
    except Exception as e:
        print("Skip", sh, e)

meta_df = pd.DataFrame(meta, columns=["sheet","header_row","score"]).sort_values("score", ascending=False)
display(meta_df.head(20))

print("Loaded candidate tables:", len(nunner_tables))
```

Auto-orientation detection + standardize to long format

This handles both formats:

weights as columns (columns are 0.5, 1, 2, 5…)

weights as rows (first column contains weights, columns are destinations/zones)

```python
def norm_token(x):
    s = str(x).strip().upper()
    s = s.replace(",", ".")
    s = s.replace("KG", "").replace("KGS", "").strip()
    return s

def parse_weight_header(x):
    s = norm_token(x)
    if re.fullmatch(r"\d+(\.\d+)?", s):
        return float(s)
    return None

def parse_weight_range_cell(x):
    s = str(x).strip().upper().replace(",", ".")
    s = re.sub(r"\s+", "", s)
    m = re.fullmatch(r"(\d+(\.\d+)?)-(\d+(\.\d+)?)", s)
    if m:
        return float(m.group(1)), float(m.group(3))
    return None

def infer_orientation(df):
    prov = {"__SOURCE_FILE","__SOURCE_SHEET","__HEADER_ROW"}
    nonprov = [c for c in df.columns if c not in prov]
    weight_cols = [c for c in nonprov if parse_weight_header(c) is not None]
    if len(weight_cols) >= 3:
        return "weights_as_columns"

    # weights-as-rows check
    first = nonprov[0] if nonprov else None
    if first is None:
        return "unknown"
    s = df[first].astype(str).head(30)
    hits = 0
    for v in s:
        v2 = str(v).strip().replace(",", ".")
        if re.fullmatch(r"\d+(\.\d+)?", v2) or parse_weight_range_cell(v) is not None:
            hits += 1
    if hits >= 5:
        return "weights_as_rows"
    return "unknown"

def melt_weights_as_columns(df):
    prov = ["__SOURCE_FILE","__SOURCE_SHEET","__HEADER_ROW"]
    nonprov = [c for c in df.columns if c not in prov]
    weight_cols = [c for c in nonprov if parse_weight_header(c) is not None]
    if len(weight_cols) < 3:
        return None

    id_cols = [c for c in nonprov if c not in weight_cols]
    # pick a likely destination column if present
    # (if multiple id cols exist, we keep them as descriptive fields)
    out = df[id_cols + weight_cols + prov].copy()
    out = out.melt(id_vars=id_cols + prov, value_vars=weight_cols, var_name="WEIGHT", value_name="RATE")
    out["WEIGHT"] = out["WEIGHT"].map(parse_weight_header)
    out["RATE"] = pd.to_numeric(out["RATE"], errors="coerce")
    out = out.dropna(subset=["WEIGHT","RATE"])
    return out

def melt_weights_as_rows(df):
    prov = ["__SOURCE_FILE","__SOURCE_SHEET","__HEADER_ROW"]
    nonprov = [c for c in df.columns if c not in prov]
    if len(nonprov) < 2:
        return None

    weight_col = nonprov[0]
    dest_cols = nonprov[1:]

    out = df[[weight_col] + dest_cols + prov].copy()
    out = out.rename(columns={weight_col: "WEIGHT_RAW"})
    out = out.melt(id_vars=["WEIGHT_RAW"] + prov, value_vars=dest_cols, var_name="DEST", value_name="RATE")
    out["RATE"] = pd.to_numeric(out["RATE"], errors="coerce")

    # parse weight raw
    w_single = out["WEIGHT_RAW"].astype(str).str.strip().str.replace(",", ".", regex=False)
    out["WEIGHT"] = pd.to_numeric(w_single, errors="coerce")

    ranges = out["WEIGHT_RAW"].map(parse_weight_range_cell)
    out["WEIGHT_FROM"] = ranges.map(lambda x: x[0] if isinstance(x, tuple) else np.nan)
    out["WEIGHT_TO"]   = ranges.map(lambda x: x[1] if isinstance(x, tuple) else np.nan)

    out = out.dropna(subset=["RATE"])
    return out

parsed = []
debug = []

for t in nunner_tables:
    sh = t["__SOURCE_SHEET"].iloc[0]
    orient = infer_orientation(t)
    debug.append((sh, orient, t.shape, list(t.columns)[:10]))

    if orient == "weights_as_columns":
        out = melt_weights_as_columns(t)
        if out is not None and len(out):
            out["PARSER"] = "weights_as_columns"
            parsed.append(out)

    elif orient == "weights_as_rows":
        out = melt_weights_as_rows(t)
        if out is not None and len(out):
            out["PARSER"] = "weights_as_rows"
            parsed.append(out)

debug_df = pd.DataFrame(debug, columns=["sheet","orientation","shape","first_cols"])
display(debug_df.head(30))

print("Parsed tables:", len(parsed))
```

Read each sheet as raw + detect the weight column position

We’ll scan rows and find where numeric weights start.

```python
def read_raw(path, sheet):
    return pd.read_excel(path, sheet_name=sheet, header=None)

def find_weight_column(df, look_rows=60, look_cols=15):
    # Find a column that has many numeric values (weights) in lower rows
    best = (None, -1)
    for c in range(min(look_cols, df.shape[1])):
        col = df.iloc[:look_rows, c]
        nums = pd.to_numeric(col.astype(str).str.replace(",", ".", regex=False), errors="coerce")
        score = nums.notna().sum()
        if score > best[1]:
            best = (c, score)
    return best  # (col_idx, score)

# quick test on the sheets showed
for sh in ["AL", "AT", "Serbia", "FR-57 Cargotec"]:
    if sh in xls.sheet_names:
        raw = read_raw(nunner_out_path, sh)
        wc, sc = find_weight_column(raw)
        print(sh, "weight_col_idx:", wc, "score:", sc)
```

Extract a clean table from each sheet (weights + numeric rate columns)

This is the key cell: it converts messy sheets into a structured dataframe.

```python
def extract_nunner_sheet(path, sheet):
    raw = read_raw(path, sheet)

    # find weight column
    weight_col, _ = find_weight_column(raw)

    # determine start row: first row where weight is numeric AND next row also numeric
    w = pd.to_numeric(raw.iloc[:, weight_col].astype(str).str.replace(",", ".", regex=False), errors="coerce")
    start_row = None
    for i in range(len(w)-1):
        if pd.notna(w.iloc[i]) and pd.notna(w.iloc[i+1]):
            start_row = i
            break
    if start_row is None:
        return None

    # slice from start row downward and keep a reasonable number of columns
    block = raw.iloc[start_row:, :].copy()
    block.columns = [f"C{c}" for c in range(block.shape[1])]

    # clean weights
    block["WEIGHT"] = pd.to_numeric(block[f"C{weight_col}"].astype(str).str.replace(",", ".", regex=False), errors="coerce")
    block = block[block["WEIGHT"].notna()].copy()

    # find numeric columns that look like rates (exclude WEIGHT col)
    numeric_cols = []
    for c in block.columns:
        if c in ["WEIGHT", f"C{weight_col}"]:
            continue
        vals = pd.to_numeric(block[c].astype(str).str.replace(",", ".", regex=False).str.replace("€","", regex=False),
                             errors="coerce")
        if vals.notna().sum() >= max(5, int(0.5 * len(block))):  # enough numeric density
            numeric_cols.append(c)

    if not numeric_cols:
        return None

    # Build long format: one row per weight per numeric column
    out = block[["WEIGHT"] + numeric_cols].copy()
    out = out.melt(id_vars=["WEIGHT"], value_vars=numeric_cols, var_name="RATE_COL", value_name="RATE_RAW")

    out["RATE"] = pd.to_numeric(
        out["RATE_RAW"].astype(str).str.replace(",", ".", regex=False).str.replace("€","", regex=False),
        errors="coerce"
    )
    out = out.dropna(subset=["RATE"])

    # Attach provenance and labels
    out["SHEET"] = sheet
    out["__SOURCE_FILE"] = Path(path).name
    out["__SOURCE_SHEET"] = sheet
    return out

parsed_sheets = []
skipped = []

for sh in rate_sheets:
    t = extract_nunner_sheet(nunner_out_path, sh)
    if t is None or len(t) == 0:
        skipped.append(sh)
    else:
        parsed_sheets.append(t)

print("Parsed sheets:", len(parsed_sheets))
print("Skipped sheets:", len(skipped))
print("Some skipped:", skipped[:20])

nunner_out_long = pd.concat(parsed_sheets, ignore_index=True)
print("nunner_out_long:", nunner_out_long.shape)
nunner_out_long.head(20)
```

A — Extract postcode bands for AT (and similar sheets)
Cell N11 — Build a “column label map” (C1 → 10-19, etc.) per sheet

This reads the header rows above the weight table and finds patterns like 10-19.

```python
import pandas as pd
import numpy as np
import re
from pathlib import Path

def get_col_band_map(path, sheet, max_rows=30):
    """
    For sheets like AT: locate the row containing postcode bands (e.g., 10-19, 20-29, ...)
    and map column index -> label.
    """
    raw = pd.read_excel(path, sheet_name=sheet, header=None, nrows=max_rows)
    # search for a row containing multiple patterns like NN-NN
    best_row = None
    best_hits = 0

    for i in range(len(raw)):
        row = raw.iloc[i].astype(str).fillna("")
        hits = row.str.contains(r"^\s*\d{2}\s*-\s*\d{2}\s*$", regex=True).sum()
        if hits > best_hits:
            best_hits = hits
            best_row = i

    if best_row is None or best_hits < 2:
        return {}  # no band row

    row = raw.iloc[best_row].astype(str).fillna("")
    band_map = {}
    for col_idx, val in enumerate(row.values):
        s = str(val).strip()
        if re.fullmatch(r"\d{2}\s*-\s*\d{2}", s):
            band_map[f"C{col_idx}"] = s.replace(" ", "")
    return band_map

# Build band maps for all parsed sheets
band_maps = {}
for sh in nunner_out_long["SHEET"].unique():
    band_maps[sh] = get_col_band_map(nunner_out_path, sh)

# Example: check AT
print("AT band map:", band_maps.get("AT", {}))
```

the sample reveals a big unit issue we should fix now before Geodis:

WEIGHT_TO values like 279.95, 335.93, 391.92 kg look plausible…

but RATE values 1.0, 2.0, 3.0… EUR are not plausible as transport costs.

That almost always means the parser accidentally picked the “blok / step / index” column as the rate (your outbound sheets had “Blok” too). For inbound, “RATE_COL” is just C? so it may be melting the wrong numeric column.

Let’s correct it analytically by choosing the best “rate column” per sheet using a simple, defensible heuristic:

Heuristic

A true rate column should have:

many numeric values

median > 5 EUR (or >10) and max reasonably high

not just 1..N integers

We’ll re-extract inbound sheets by detecting:

weight column = numeric column with monotonic-ish increasing breakpoints

rate column = numeric column with realistic magnitudes

Fix Nunner INBOUND: pick the correct RATE column

Re-extract each inbound sheet with “best rate column” logic

```python
import pandas as pd
import numpy as np
import re
from pathlib import Path

nunner_in_path = FILES["nunner_in"]
xls = pd.ExcelFile(nunner_in_path)

def read_raw(path, sheet):
    return pd.read_excel(path, sheet_name=sheet, header=None)

def numeric_series(s):
    return pd.to_numeric(s.astype(str).str.replace(",", ".", regex=False).str.replace("€","", regex=False),
                         errors="coerce")

def find_best_weight_and_rate_cols(raw, max_rows=120, max_cols=20):
    # evaluate first max_rows rows
    raw2 = raw.iloc[:max_rows, :max_cols].copy()

    col_stats = []
    for c in raw2.columns:
        nums = numeric_series(raw2[c])
        nn = nums.notna().sum()
        if nn < 8:
            continue
        col_stats.append((c, nn, nums))

    if not col_stats:
        return None, None

    # Choose WEIGHT column: numeric column with many values and (often) increasing
    def weight_score(nums):
        v = nums.dropna().values
        if len(v) < 8:
            return -1
        inc_ratio = (np.diff(v) >= 0).mean() if len(v) > 1 else 0
        # weights usually have larger values than 1..N and tend to increase
        return len(v) + 10 * inc_ratio + (np.nanmax(v) / 1000)

    weight_candidates = sorted(col_stats, key=lambda t: weight_score(t[2]), reverse=True)
    weight_col = weight_candidates[0][0]
    w_nums = weight_candidates[0][2]

    # Choose RATE column: numeric column not equal weight_col with realistic magnitude
    def rate_score(nums):
        v = nums.dropna().values
        if len(v) < 8:
            return -1
        med = np.median(v)
        mx = np.max(v)
        # penalize "1..N" integer index columns
        looks_like_index = (np.all(np.isclose(v, np.round(v))) and mx <= 50 and med <= 10)
        score = len(v)
        score += 5 * (med > 10) + 3 * (mx > 50) + 2 * (mx > 200)
        score -= 10 * looks_like_index
        return score

    rate_candidates = []
    for c, nn, nums in col_stats:
        if c == weight_col:
            continue
        rate_candidates.append((c, rate_score(nums), nums))

    rate_candidates = sorted(rate_candidates, key=lambda x: x[1], reverse=True)
    rate_col = rate_candidates[0][0] if rate_candidates and rate_candidates[0][1] > 0 else None

    return weight_col, rate_col

def extract_inbound_sheet_best(path, sheet):
    raw = read_raw(path, sheet)
    wcol, rcol = find_best_weight_and_rate_cols(raw)
    if wcol is None or rcol is None:
        return None

    # find start row where weight becomes numeric for 2 consecutive rows
    w = numeric_series(raw[wcol])
    start_row = None
    for i in range(len(w)-1):
        if pd.notna(w.iloc[i]) and pd.notna(w.iloc[i+1]):
            start_row = i
            break
    if start_row is None:
        return None

    block = raw.iloc[start_row:, [wcol, rcol]].copy()
    block.columns = ["WEIGHT", "RATE"]
    block["WEIGHT"] = numeric_series(block["WEIGHT"])
    block["RATE"] = numeric_series(block["RATE"])
    block = block.dropna(subset=["WEIGHT","RATE"])

    block["SHEET"] = sheet
    block["__SOURCE_FILE"] = Path(path).name
    block["__SOURCE_SHEET"] = sheet
    block["WEIGHT_COL_IDX"] = int(wcol)
    block["RATE_COL_IDX"] = int(rcol)
    return block

# Run extraction
parsed = []
skipped = []

for sh in xls.sheet_names:
    if "REMARK" in sh.upper():
        continue
    t = extract_inbound_sheet_best(nunner_in_path, sh)
    if t is None or len(t) == 0:
        skipped.append(sh)
    else:
        parsed.append(t)

print("Parsed:", len(parsed), "Skipped:", len(skipped))
print("Some skipped:", skipped[:10])

in_long2 = pd.concat(parsed, ignore_index=True)
print("in_long2:", in_long2.shape)
in_long2.head(10)
```

Banding + lane parsing + final inbound table (v2)

```python
df = in_long2.copy()
df["CARRIER"] = "NUNNER"
df["DIRECTION"] = "INBOUND"
df["WEIGHT_UOM"] = "KG"
df["CURRENCY"] = "EUR"

# half-open bands per sheet
def bands(g):
    g = g.sort_values("WEIGHT").copy()
    g = g.drop_duplicates(subset=["WEIGHT"])
    g["WEIGHT_FROM"] = g["WEIGHT"].shift(1).fillna(0)
    g["WEIGHT_TO"] = g["WEIGHT"]
    return g[g["WEIGHT_TO"] > g["WEIGHT_FROM"]]

df_b = df.groupby(["SHEET"], dropna=False, group_keys=False).apply(bands)

def parse_lane(sheet):
    s = sheet.upper()
    m = re.search(r"([A-Z]{2}).*?(?:TO|->|–|-|>|_).*?([A-Z]{2})", s.replace(" ", ""))
    if m:
        return m.group(1), m.group(2)
    return (np.nan, np.nan)

lanes = df_b["SHEET"].apply(parse_lane)
df_b["ORIGIN_COUNTRY"] = lanes.map(lambda x: x[0])
df_b["DESTINATION_COUNTRY"] = lanes.map(lambda x: x[1])
df_b["LANE_LABEL"] = df_b["SHEET"]

nunner_in_fact_final_v2 = df_b[
    ["CARRIER","DIRECTION","ORIGIN_COUNTRY","DESTINATION_COUNTRY","LANE_LABEL",
     "WEIGHT_FROM","WEIGHT_TO","WEIGHT_UOM","RATE","CURRENCY",
     "WEIGHT_COL_IDX","RATE_COL_IDX","__SOURCE_FILE","__SOURCE_SHEET"]
].copy()

print("✅ Nunner IN final v2:", nunner_in_fact_final_v2.shape)
display(nunner_in_fact_final_v2.head(5))

nunner_in_fact_final_v2.to_csv("nunner_in_rate_fact_final_v2.csv", index=False)
print("Saved: nunner_in_rate_fact_final_v2.csv")
```

```python
import pandas as pd

def ensure_df(name, csv_path):
    if name in globals():
        return globals()[name]
    return pd.read_csv(csv_path)

nunner_out = ensure_df("nunner_out_rate_fact_final_v2", "nunner_out_rate_fact_final_v2.csv")
nunner_in  = ensure_df("nunner_in_rate_fact_final_v2",  "nunner_in_rate_fact_final_v2.csv")
geodis     = ensure_df("geodis_rate_fact_final",        "geodis_rate_fact_final.csv")

print("nunner_out:", nunner_out.shape)
print("nunner_in :", nunner_in.shape)
print("geodis    :", geodis.shape)
```

#### Harmonize rate-card tables into one matching universe

```python
# We’ll keep only the columns we need and align naming.

def prep_rates(df, carrier, direction):
    out = df.copy()
    out["CARRIER"] = carrier
    out["DIRECTION"] = direction
    # normalize country columns if present
    for c in ["ORIGIN_COUNTRY","DESTINATION_COUNTRY"]:
        if c in out.columns:
            out[c] = out[c].astype(str).str.upper().str.strip()
    return out

rates_out = prep_rates(nunner_out, "NUNNER", "OUTBOUND")
rates_in  = prep_rates(nunner_in,  "NUNNER", "INBOUND")
rates_geo = prep_rates(geodis,     "GEODIS", "OUTBOUND")

# Ensure these columns exist (even if empty) so we can concatenate safely
for r in [rates_out, rates_in, rates_geo]:
    for col in ["ORIGIN_COUNTRY","DESTINATION_COUNTRY","SERVICE_LEVEL"]:
        if col not in r.columns:
            r[col] = pd.NA

rate_universe = pd.concat([rates_out, rates_in, rates_geo], ignore_index=True)

rate_universe[["CARRIER","DIRECTION"]].value_counts()
```

#### Matching logic (carrier-specific)

* Nunner OUTBOUND

Match on: destination country + weight band

* Nunner INBOUND

Match on: origin + destination + weight band

* Geodis OUTBOUND

Match on: service level (AFF → sheet contains AFF) + weight band
(If no service match is possible, we fall back to weight-only for Geodis, but we’ll measure how often.)

```python
import numpy as np

def match_weight_bands(ship_df, rate_df, keys, weight_col="BILLABLE_WEIGHT"):
    m = ship_df.merge(rate_df, on=keys, how="left", suffixes=("", "_RATE"))
    m = m[(m[weight_col] >= m["WEIGHT_FROM"]) & (m[weight_col] < m["WEIGHT_TO"])].copy()
    return m

# --- NUNNER OUTBOUND ---
lh_nunner_out = lh[(lh["CARRIER"]=="NUNNER") & (lh["DIRECTION"]=="OUTBOUND")].copy()
rc_nunner_out = rate_universe[(rate_universe["CARRIER"]=="NUNNER") & (rate_universe["DIRECTION"]=="OUTBOUND")].copy()

matched_nunner_out = match_weight_bands(
    lh_nunner_out,
    rc_nunner_out,
    keys=["CARRIER","DIRECTION","DESTINATION_COUNTRY"]
)

# --- NUNNER INBOUND ---
lh_nunner_in = lh[(lh["CARRIER"]=="NUNNER") & (lh["DIRECTION"]=="INBOUND")].copy()
rc_nunner_in = rate_universe[(rate_universe["CARRIER"]=="NUNNER") & (rate_universe["DIRECTION"]=="INBOUND")].copy()

matched_nunner_in = match_weight_bands(
    lh_nunner_in,
    rc_nunner_in,
    keys=["CARRIER","DIRECTION","ORIGIN_COUNTRY","DESTINATION_COUNTRY"]
)

# --- GEODIS OUTBOUND ---
lh_geodis = lh[(lh["CARRIER"]=="GEODIS") & (lh["DIRECTION"]=="OUTBOUND")].copy()
rc_geodis = rate_universe[(rate_universe["CARRIER"]=="GEODIS") & (rate_universe["DIRECTION"]=="OUTBOUND")].copy()

# Service filter: if shipment service contains "AFF", prefer rate rows from sheets containing "AFF"
lh_geodis["SERVICE_IS_AFF"] = lh_geodis["SERVICE_LEVEL"].str.contains("AFF", na=False)

rc_geodis_aff = rc_geodis[rc_geodis["SOURCE_SHEET_UP"].str.contains("AFF", na=False)].copy()
rc_geodis_any = rc_geodis.copy()

matched_geodis_aff = match_weight_bands(
    lh_geodis[lh_geodis["SERVICE_IS_AFF"]].copy(),
    rc_geodis_aff,
    keys=["CARRIER","DIRECTION"]
)

matched_geodis_fallback = match_weight_bands(
    lh_geodis[~lh_geodis["SERVICE_IS_AFF"]].copy(),
    rc_geodis_any,
    keys=["CARRIER","DIRECTION"]
)

# Combine all matches
matched_all = pd.concat([matched_nunner_out, matched_nunner_in, matched_geodis_aff, matched_geodis_fallback], ignore_index=True)

matched_all.shape
```

####Resolve duplicates (if multiple rates match a shipment)

Sometimes a shipment can match multiple tariff rows (e.g., overlapping tables).
We’ll keep the lowest rate as a conservative baseline (and we’ll count how often this happens).

```python
# Define a unique shipment key for dedup
matched_all["SHIP_KEY"] = (
    matched_all["Shipment Id"].astype(str) + "|" +
    matched_all["Invoice No"].astype(str) + "|" +
    matched_all["CARRIER"].astype(str)
)

dup_counts = matched_all["SHIP_KEY"].value_counts()
print("Shipments with >1 matched rate:", (dup_counts > 1).sum())

# Keep cheapest matched rate per shipment
best_match = (
    matched_all.sort_values(["SHIP_KEY", "RATE"], ascending=[True, True])
              .drop_duplicates("SHIP_KEY", keep="first")
              .copy()
)

best_match.shape
```

####Compute baseline expected cost and errors

```python
# Simulation helpers

import numpy as np
import pandas as pd

def build_min_rate_lookup(rate_df, w_from="WEIGHT_FROM", w_to="WEIGHT_TO", rate="RATE", max_w=None):
    """
    Build a per-weight lookup array where price[w] = min RATE among all intervals containing w.
    This handles overlapping intervals by taking the minimum.
    """
    r = rate_df.copy()
    r[w_from] = pd.to_numeric(r[w_from], errors="coerce")
    r[w_to]   = pd.to_numeric(r[w_to], errors="coerce")
    r[rate]   = pd.to_numeric(r[rate], errors="coerce")

    r = r.dropna(subset=[w_from, w_to, rate]).copy()
    r = r.sort_values([w_from, w_to]).reset_index(drop=True)

    # Decide maximum weight to cover
    if max_w is None:
        max_w = int(np.ceil(r[w_to].max()))
    max_w = int(max_w)

    # Initialize with +inf then take min
    price = np.full(max_w + 1, np.inf, dtype=float)

    for wf, wt, rr in r[[w_from, w_to, rate]].itertuples(index=False, name=None):
        wf_i = int(np.floor(wf))
        wt_i = int(np.ceil(wt))
        if wt_i <= 0:
            continue
        wf_i = max(wf_i, 0)
        wt_i = min(wt_i, max_w + 1)
        if wf_i < wt_i:
            price[wf_i:wt_i] = np.minimum(price[wf_i:wt_i], rr)

    # Replace inf with NaN (uncovered weights)
    price[np.isinf(price)] = np.nan
    return price

def quote_with_lookup(ship_df, weight_col, lookup, out_col):
    """
    Apply per-weight lookup to shipments. Weight is floored to int kg.
    """
    out = ship_df[[COL_KEY, weight_col]].copy()
    w = pd.to_numeric(out[weight_col], errors="coerce")
    w_int = np.floor(w).astype("Int64")  # keeps NaNs

    def _get(v):
        if pd.isna(v):
            return np.nan
        v = int(v)
        if v < 0:
            return np.nan
        if v >= len(lookup):
            return lookup[-1]  # cap at last
        return lookup[v]

    out[out_col] = w_int.map(_get).astype(float)
    return out[[COL_KEY, out_col]]

def _to_num(s):
    """Robust numeric conversion (handles strings, commas, blanks)."""
    return pd.to_numeric(
        s.astype(str).str.replace(",", "", regex=False).str.strip(),
        errors="coerce"
    )

def _prep_rate_table(rate_df: pd.DataFrame,
                     w_from_col="WEIGHT_FROM",
                     w_to_col="WEIGHT_TO",
                     rate_col="RATE") -> pd.DataFrame:
    """
    Clean + standardize a weight-break rate table.
    Also deduplicates identical intervals to avoid IntervalIndex overlap errors.
    """
    r = rate_df.copy()

    missing = [c for c in [w_from_col, w_to_col, rate_col] if c not in r.columns]
    if missing:
        raise KeyError(f"Rate table missing required columns: {missing}")

    r[w_from_col] = _to_num(r[w_from_col])
    r[w_to_col]   = _to_num(r[w_to_col])
    r[rate_col]   = _to_num(r[rate_col])

    r = r.dropna(subset=[w_from_col, rate_col]).copy()
    r[w_to_col] = r[w_to_col].fillna(np.inf)

    # remove duplicate intervals
    # Keep the minimum rate for the same interval
    r = (
        r.groupby([w_from_col, w_to_col], as_index=False)[rate_col]
         .min()
         .sort_values([w_from_col, w_to_col])
         .reset_index(drop=True)
    )

    # Ensure last break covers everything
    r.loc[r.index.max(), w_to_col] = np.inf

    return r


def reprice(shipments: pd.DataFrame,
            rate_df: pd.DataFrame,
            weight_col="PRICING_WEIGHT",
            w_from_col="WEIGHT_FROM",
            w_to_col="WEIGHT_TO",
            rate_col="RATE",
            out_col="sim_rate") -> pd.DataFrame:
    """
    Assigns sim_rate by matching PRICING_WEIGHT into WEIGHT_FROM/WEIGHT_TO breaks.
    - Uses IntervalIndex lookup (fast + reliable).
    - Closed='left' matches [from, to) and last break is [last_from, inf).
    - Adds a boolean flag _sim_unmatched for diagnostics.
    """
    out = shipments.copy()

    if weight_col not in out.columns:
        raise KeyError(f"{weight_col} not found in shipments. Create it before repricing.")

    out[weight_col] = _to_num(out[weight_col])
    r = _prep_rate_table(rate_df, w_from_col, w_to_col, rate_col)

    # Build intervals and map weights to breaks
    intervals = pd.IntervalIndex.from_arrays(r[w_from_col], r[w_to_col], closed="left")
    idx = intervals.get_indexer(out[weight_col].to_numpy())

    rates_arr = r[rate_col].to_numpy()
    out[out_col] = np.where(idx >= 0, rates_arr[idx], np.nan)
    out["_sim_unmatched"] = idx < 0

    return out

def add_savings(df: pd.DataFrame,
                sim_rate_col="sim_rate",
                invoiced_col="invoiced_linehaul_amount",
                expected_col="expected_linehaul_cost") -> pd.DataFrame:
    """
    Adds delta/savings columns safely.
    - delta: invoiced - sim_rate (positive means simulated cheaper)
    - sim_savings: expected - sim_rate (if expected is your baseline model)
    """
    out = df.copy()

    for c in [sim_rate_col, invoiced_col, expected_col]:
        if c in out.columns:
            out[c] = _to_num(out[c])

    if sim_rate_col in out.columns and invoiced_col in out.columns:
        out["delta"] = out[invoiced_col] - out[sim_rate_col]
    else:
        out["delta"] = np.nan

    if sim_rate_col in out.columns and expected_col in out.columns:
        out["sim_savings"] = out[expected_col] - out[sim_rate_col]
    else:
        out["sim_savings"] = np.nan

    return out

def diagnose_sim(df: pd.DataFrame, weight_col="PRICING_WEIGHT", sim_rate_col="sim_rate", key_cols=None, head=15):
    """Quick visibility into why sim_rate might be NaN."""
    if key_cols is None:
        key_cols = [c for c in ["SHIP_KEY", weight_col, "BILLABLE_WEIGHT", "invoiced_linehaul_amount", sim_rate_col] if c in df.columns]

    print("Rows:", len(df))
    if weight_col in df.columns:
        print(f"{weight_col} NaN %:", round(df[weight_col].isna().mean(), 4),
              "| min:", df[weight_col].min(),
              "| max:", df[weight_col].max())

    if sim_rate_col in df.columns:
        print(f"{sim_rate_col} NaN %:", round(df[sim_rate_col].isna().mean(), 4),
              "| min:", df[sim_rate_col].min(),
              "| max:", df[sim_rate_col].max())

    if "_sim_unmatched" in df.columns and df["_sim_unmatched"].any():
        bad = df[df["_sim_unmatched"]].copy()
        print("Unmatched rows:", len(bad))
        display(bad[key_cols].head(head))


```

```python

#  Quote each shipment under each carrier
# Why: to evaluate any threshold rule, we need the "what would it cost?" for each carrier.

import pandas as pd
import numpy as np

def _quote_weight_only(ship_df, rate_df, weight_col, out_col):
    # Uses the validated reprice() helper if available; otherwise does a simple interval lookup.
    if "reprice" in globals():
        tmp = reprice(ship_df, rate_df, weight_col=weight_col, out_col=out_col)
        return tmp[[COL_KEY, out_col]]
    # fallback
    r = rate_df.copy()
    r["WEIGHT_FROM"] = pd.to_numeric(r["WEIGHT_FROM"], errors="coerce")
    r["WEIGHT_TO"]   = pd.to_numeric(r["WEIGHT_TO"], errors="coerce").fillna(np.inf)
    r["RATE"]        = pd.to_numeric(r["RATE"], errors="coerce")
    r = r.dropna(subset=["WEIGHT_FROM","RATE"]).sort_values(["WEIGHT_FROM","WEIGHT_TO"])
    intervals = pd.IntervalIndex.from_arrays(r["WEIGHT_FROM"], r["WEIGHT_TO"], closed="left")
    w = pd.to_numeric(ship_df[weight_col], errors="coerce")
    idx = intervals.get_indexer(w)
    ship_df = ship_df.copy()
    ship_df[out_col] = np.where(idx>=0, r["RATE"].to_numpy()[idx], np.nan)
    return ship_df[[COL_KEY, out_col]]

q_geo = _quote_weight_only(SHIP, RATES_GEODIS, COL_WEIGHT, "rate_geodis")

# --- DHL (SPECIAL CASE: overlapping intervals) ---
max_w_ship = int(
    np.ceil(pd.to_numeric(SHIP[COL_WEIGHT], errors="coerce").max())
)

dhl_lookup = build_min_rate_lookup(
    RATES_DHL,
    w_from="WEIGHT_FROM",
    w_to="WEIGHT_TO",
    rate="RATE",
    max_w=max_w_ship
)

q_dhl = quote_with_lookup(
    SHIP,
    COL_WEIGHT,
    dhl_lookup,
    "rate_dhl"
)

# NUNNER also overlaps
nun_lookup = build_min_rate_lookup(
    RATES_NUNNER,
    w_from="WEIGHT_FROM",
    w_to="WEIGHT_TO",
    rate="RATE",
    max_w=max_w_ship
)

q_nun = quote_with_lookup(
    SHIP,
    COL_WEIGHT,
    nun_lookup,
    "rate_nunner"
)


quotes = (
    SHIP
    .merge(q_geo, on=COL_KEY, how="left")
    .merge(q_dhl, on=COL_KEY, how="left")
    .merge(q_nun, on=COL_KEY, how="left")
)

print(
    "Quote NaN %:",
    "GEODIS", round(quotes["rate_geodis"].isna().mean(), 4),
    "| DHL", round(quotes["rate_dhl"].isna().mean(), 4),
    "| NUNNER", round(quotes["rate_nunner"].isna().mean(), 4)
)

```

```python
# Baseline policy: 150kg rule + fallback
# Why: create a realistic "current policy" benchmark to compare against optimal thresholds.

import numpy as np
import pandas as pd

def baseline_150_with_fallback(r):
    w = r[COL_WEIGHT]

    # Primary rule
    primary = "DHL" if w < 150 else "NUNNER"

    # Try primary if rate exists
    if primary == "DHL" and pd.notna(r["rate_dhl"]):
        return "DHL", r["rate_dhl"]
    if primary == "NUNNER" and pd.notna(r["rate_nunner"]):
        return "NUNNER", r["rate_nunner"]

    # Fallback order (business decision): GEODIS then NUNNER then DHL
    if pd.notna(r["rate_geodis"]):
        return "GEODIS", r["rate_geodis"]
    if pd.notna(r["rate_nunner"]):
        return "NUNNER", r["rate_nunner"]
    if pd.notna(r["rate_dhl"]):
        return "DHL", r["rate_dhl"]

    return None, np.nan

baseline = quotes.copy()
baseline[["carrier_baseline", "cost_baseline"]] = baseline.apply(
    lambda r: pd.Series(baseline_150_with_fallback(r)), axis=1
)

print("Baseline missing cost %:", round(baseline["cost_baseline"].isna().mean(), 4))
print("Baseline carrier mix:")
display(baseline["carrier_baseline"].value_counts(dropna=False).to_frame("count"))

print("Baseline total cost:", round(baseline["cost_baseline"].sum(skipna=True), 2))
print("Baseline avg cost:", round(baseline["cost_baseline"].mean(skipna=True), 2))
```

```python
# Threshold search (global)
# Why: test alternative breaks and find a threshold that minimizes total cost (with the same fallback logic).

candidates = list(range(50, 401, 10))  # 50..400 in 10kg steps

def cost_under_threshold(r, t):
    w = r[COL_WEIGHT]
    primary = "DHL" if w < t else "NUNNER"

    if primary == "DHL" and pd.notna(r["rate_dhl"]):
        return r["rate_dhl"]
    if primary == "NUNNER" and pd.notna(r["rate_nunner"]):
        return r["rate_nunner"]

    # same fallback as baseline
    if pd.notna(r["rate_geodis"]):
        return r["rate_geodis"]
    if pd.notna(r["rate_nunner"]):
        return r["rate_nunner"]
    if pd.notna(r["rate_dhl"]):
        return r["rate_dhl"]

    return np.nan

rows = []
for t in candidates:
    c = quotes.apply(lambda r: cost_under_threshold(r, t), axis=1)
    rows.append({
        "threshold_kg": t,
        "total_cost": float(np.nansum(c.values)),
        "avg_cost": float(np.nanmean(c.values)),
        "missing_cost_pct": float(pd.isna(c).mean())
    })

opt = pd.DataFrame(rows).sort_values("total_cost")
display(opt.head(10))

best_t = int(opt.iloc[0]["threshold_kg"])
best_total = float(opt.iloc[0]["total_cost"])
baseline_total = float(baseline["cost_baseline"].sum(skipna=True))

print("Best threshold:", best_t)
print("Baseline total cost (150kg rule):", round(baseline_total, 2))
print("Best total cost:", round(best_total, 2))
print("Difference (best - baseline):", round(best_total - baseline_total, 2))
```

```python
# --- Upgrade A: Identify the cheapest carrier by bin + highlight crossovers ---
# Why: makes the output directly interpretable as "where should we switch?"

s = summary.copy()

# Differences (positive => DHL more expensive)
s["dhl_minus_nunner"] = s["dhl_avg"] - s["nunner_avg"]

# Winner among DHL vs NUNNER where both exist
def _winner(row):
    if pd.isna(row["dhl_avg"]) and pd.isna(row["nunner_avg"]):
        return "NO_QUOTE"
    if pd.isna(row["dhl_avg"]):
        return "NUNNER"
    if pd.isna(row["nunner_avg"]):
        return "DHL"
    return "DHL" if row["dhl_avg"] < row["nunner_avg"] else "NUNNER"

s["winner_dhl_vs_nunner"] = s.apply(_winner, axis=1)

# Crossover detection: where winner changes across adjacent bins
s["winner_change"] = s["winner_dhl_vs_nunner"].ne(s["winner_dhl_vs_nunner"].shift(1))

display(s[[
    "W_BIN", "n",
    "dhl_avg", "nunner_avg", "dhl_minus_nunner",
    "winner_dhl_vs_nunner", "winner_change",
    "dhl_nan_pct"
]])
```

### Upgrade B: Add robustness (median + percentiles)

Why we’re doing this: Averages can be misleading if there are outliers or if the rate table isn’t lane-filtered yet. Median/percentiles show whether the “winner” is consistent.

```python
# --- Upgrade B: Robust stats (median + 25/75 pct) ---
# Why: avoids conclusions driven by a few extreme prices

def q25(x):
    return float(np.nanpercentile(x, 25))
def q75(x):
    return float(np.nanpercentile(x, 75))

robust = tmp.groupby("W_BIN", observed=False).agg(
    n=("W_BIN","size"),

    dhl_med=("rate_dhl","median"),
    dhl_p25=("rate_dhl", q25),
    dhl_p75=("rate_dhl", q75),
    dhl_nan_pct=("rate_dhl", lambda s: float(s.isna().mean())),

    nun_med=("rate_nunner","median"),
    nun_p25=("rate_nunner", q25),
    nun_p75=("rate_nunner", q75),

    geo_med=("rate_geodis","median")
).reset_index()

# Winner using medians
robust["winner_by_median"] = np.where(
    (robust["dhl_med"].notna()) & (robust["nun_med"].notna()) & (robust["dhl_med"] < robust["nun_med"]),
    "DHL",
    np.where(
        (robust["dhl_med"].isna()) & (robust["nun_med"].notna()),
        "NUNNER",
        np.where(
            (robust["nun_med"].isna()) & (robust["dhl_med"].notna()),
            "DHL",
            "NUNNER"
        )
    )
)

display(robust)
```
